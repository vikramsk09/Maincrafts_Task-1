3. Analysis Questions

```python
# Average final grade (G3)
average = df['G3'].mean()
print(f"Average final grade: {average:.2f}")
```

```python
# How many students scored above 15?
scored_above_15 = (df['G3'] > 15).sum()
print(f"Total number of students who scored above 15 are - {scored_above_15}")
```

```python
# Correlation between study time and G3 (performance)
correlation = df['studytime'].corr(df['G3'])
print(f"Corrlation between studytime and final grade (G3): {correlation:.2f}")
```

Conclusion: There is a good +ve correlation of 0.25 between these 2 columns

```python
# Which gender performs better on average?
average = df.groupby('sex')['G3'].mean()
print(average)
```

Conclusion: Females performed better on average (as compared to Males)

4. Visualizations

```python
# Histogram of grades
sns.histplot(df['G3'])

plt.title("Distribution of Final Grades")
plt.xlabel("G3 Score")
plt.ylabel("Count")

plt.show()
```

```python
# Scatterplot: Study time vs. Grades
sns.scatterplot(x='studytime', y='G3', data=df)

plt.title('Study Time vs Final Grades')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
# Bar Chart: Male vs. Female average score
average_score = df.groupby('sex')['G3'].mean().reset_index()

sns.barplot(x='sex', y='G3', data=average_score)

plt.title('Male vs Female Average Score')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')

plt.show()
```

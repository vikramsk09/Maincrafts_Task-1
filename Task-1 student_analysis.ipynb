3. Analysis Questions

```python
# Average final grade (G3)
average = df['G3'].mean()
print(f"Average of final grade (G3) - {average:.2f}")
```

```python
# Students who scored above 15
scored_above_15 = (df['G3'] > 15).sum()
print(f"There are total {scored_above_15} students who had scored above 15 (final grades).")
```

Correlation between 'studytime' and 'G3' (performance)

- 1 - Strong +ve correlation
- -1 - Strong -ve correlation
- 0 - No correlation

```python
# Correlation between study time and performance
correlation = df['studytime'].corr(df['G3'])
print(f"Correlation between studytime and final grade (G3): {correlation:.2f}")
```

Conclusion: As the correlation is only 0.10, this shows that there is a very little +ve correlation

```python
# Checking which gender performs better on average
df.groupby('sex')['G3'].mean()
```

Conclusion: Males performed better on average (as compared to Females)

4. Visualizations

```python
# Histogram of grades (G3)
sns.histplot(df['G3'])
plt.title('Distribution of Final Grades')
plt.xlabel('G3 Score')
plt.ylabel('Count')

plt.show()
```

```python
# Scatterplot: Studytime vs. Grades
sns.scatterplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grades')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
# Bar chart: Male vs. Female average scores
avg_scores = df.groupby('sex')['G3'].mean().reset_index()

sns.barplot(x='sex', y='G3', data=avg_scores)
plt.title('Male vs Female Average Score')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')

plt.show()
```
